```python
import pandas as pd
import re
import os

def filter_excel_by_pattern(input_filepath, output_filepath, pattern):
    """
    Reads an Excel file, filters rows based on a regex pattern in any cell,
    and saves the matching rows to a new Excel file.

    Args:
        input_filepath (str): The path to the input Excel file (.xlsx).
        output_filepath (str): The path where the output Excel file will be saved.
        pattern (str): The regular expression pattern to search for.
    """
    # --- 1. Input Validation ---
    if not os.path.exists(input_filepath):
        print(f"Error: The file '{input_filepath}' was not found.")
        return

    print(f"Reading data from '{input_filepath}'...")

    try:
        # --- 2. Read Excel File ---
        # We use header=None because the column with the pattern might not have a name
        # or we don't know the name.
        df = pd.read_excel(input_filepath, header=None, engine='openpyxl')
    except Exception as e:
        print(f"An error occurred while reading the Excel file: {e}")
        return

    # --- 3. Prepare for Regex Search ---
    # Compile the regex for efficiency
    regex = re.compile(pattern)

    # Convert the entire DataFrame to string type to ensure regex can be applied
    # to all cells, regardless of their original data type (e.g., numbers, dates).
    df_str = df.astype(str)

    # --- 4. Apply the Filter ---
    # The core logic:
    # - df_str.applymap(lambda x: bool(regex.search(x))): Applies the regex search to every single cell (element-wise).
    #   Returns a DataFrame of True/False values.
    # - .any(axis=1): For each row, checks if there is at least one `True` value.
    #   This gives us a boolean Series (mask) indicating which rows to keep.
    mask = df_str.applymap(lambda x: bool(regex.search(x))).any(axis=1)

    # Use the mask to select the matching rows from the original DataFrame
    result_df = df[mask]

    # --- 5. Save the Output ---
    if not result_df.empty:
        print(f"Found {len(result_df)} matching rows.")
        try:
            # Save the filtered data to a new Excel file.
            # index=False and header=False ensure we only write the data,
            # without pandas' default row numbers or column headers.
            result_df.to_excel(output_filepath, index=False, header=False, engine='openpyxl')
            print(f"Successfully created filtered file at '{output_filepath}'")
        except Exception as e:
            print(f"An error occurred while saving the output file: {e}")
    else:
        print("No rows found matching the specified pattern.")


if __name__ == '__main__':
    # --- Configuration ---
    # The pattern looks for a string that starts with 2 digits, followed by 3 letters,
    # the number 1, and ends with 4 digits. \b denotes a word boundary.
    registration_number_pattern = r'\b\d{2}[A-Za-z]{3}1\d{4}\b'

    # --- Get input from the user ---
    input_file = input("Enter the name of the input Excel file (e.g., input_data.xlsx): ")
    output_file = input("Enter the name for the output Excel file (e.g., filtered_output.xlsx): ")

    # --- Run the main function ---
    filter_excel_by_pattern(
        input_filepath=input_file,
        output_filepath=output_file,
        pattern=registration_number_pattern
    )

```
